# Multi-Gram-Implimentation

```python
Multigram = 1
```

```python
def generate_multigram_tokens(tokens, Multigram):
    return [' '.join(tokens[i:i + Multigram]) for i in range(len(tokens) - Multigram + 1)]
```

```python
MultiGramTokens_Train = []
MultiGramTokens_Test = []
```

```python
print(len(MultiGramTokens_Train))
```

```python
def create_multigram_tokens_for_training(Training_data):
    Training_tokens = []

    for category in Training_data:
        for document in Training_data[category]:

            filtered_text = document.translate(str.maketrans('', '', string.punctuation))
            filtered_text = filtered_text.lower()

            tokens = filtered_text.split()
            Training_tokens.extend(tokens)

            bigrams = generate_multigram_tokens(tokens, Multigram)
            Training_tokens.extend(bigrams)

    return Training_tokens
```

```python
MultiGramTokens_Train = create_multigram_tokens_for_training(Training_data)
```

```python
print(len(MultiGramTokens_Train))
```

```python
def create_stop_words_and_unique_tokens(training_tokens, testing_tokens, top_n=200):

    frequent_word_count = {}
    for word in training_tokens:
        frequent_word_count[word] = frequent_word_count.get(word, 0) + 1

    frequent_word_list = [(word, count) for word, count in frequent_word_count.items()]

    frequent_word_list.sort(key=lambda x: x[1], reverse=True)

    stop_words = [word for word, count in frequent_word_list[:top_n]]

    unique_training_tokens = set(training_tokens)
    unique_testing_tokens = set(testing_tokens)

    multigram_unique_training_tokens_after_removing_stop_words = [
        token for token in unique_training_tokens if token not in stop_words
    ]

    unique_testing_tokens_after_removing_stop_words = [
        token for token in unique_testing_tokens if token not in stop_words
    ]

    return stop_words, multigram_unique_training_tokens_after_removing_stop_words

```

```python
def count_each_category_tokens(Training_data):
    Token_For_Each_Category = {}
    Each_Token_Count = {}

    for category, document in Training_data.items():
        Each_Token_Count[category] = {}
        Token_For_Each_Category[category] = 0

        for doc in document:
            filtered_text = doc.translate(str.maketrans('', '', string.punctuation))
            filtered_text = filtered_text.lower()
            tokens = filtered_text.split()
            multigram = generate_multigram_tokens(tokens, Multigram)
            for token in multigram:
                if token not in Each_Token_Count[category]:
                    Each_Token_Count[category][token] = 0
                Each_Token_Count[category][token] += 1
                Token_For_Each_Category[category] += 1

    return Token_For_Each_Category, Each_Token_Count
```

```python
def calculate_likelihood(Each_Token_Count, Unique_Training_Tokens_After_Removing_Stop_Words, Token_For_Each_Category, voc_size_M):
    Likeihood_For_Each_Token = {}

    for category in Each_Token_Count:
        Likeihood_For_Each_Token[category] = {}

        for token in Unique_Training_Tokens_After_Removing_Stop_Words:

            Likeihood_For_Each_Token[category][token] = math.log(
                (Each_Token_Count[category].get(token, 0) + 0.01) /
                (Token_For_Each_Category[category] + voc_size_M)
            )

    return Likeihood_For_Each_Token
```

```python
def predict_category(Testing_data, Prior_Probability, Likelihood_For_Each_Token, Stop_words, gram):
    True_Predicted_Category = {}
    True_Predicted_Category_Count = 0
    Total_Predicted_Category_Count = 0

    for test_category, data in Testing_data.items():
        for document in data:
            filtered_text = document.translate(str.maketrans('', '', string.punctuation))
            filtered_text = filtered_text.lower()
            Testing_tokens = filtered_text.split()
            Testing_tokens_prediction = generate_multigram_tokens(Testing_tokens, Multigram)

            Prediction_For_Each_Category = {}

            Unique_test_token = set(Testing_tokens_prediction)
            Unique_test_token_after_removing_stop_word = [token for token in Unique_test_token if token not in Stop_words]

            for category in Prior_Probability:
                Prediction_For_Each_Category[category] = Prior_Probability[category]
                for token in Unique_test_token_after_removing_stop_word:
                    if token in Likelihood_For_Each_Token[category]:
                        Prediction_For_Each_Category[category] += Likelihood_For_Each_Token[category][token]

            Bayes_Predicted_Category = max(Prediction_For_Each_Category, key=Prediction_For_Each_Category.get)
            if Bayes_Predicted_Category == test_category:
                True_Predicted_Category[test_category] = True_Predicted_Category.get(test_category, 0) + 1
                True_Predicted_Category_Count += 1

            Total_Predicted_Category_Count += 1

    accuracy = True_Predicted_Category_Count / Total_Predicted_Category_Count
    return accuracy, True_Predicted_Category, True_Predicted_Category_Count, Total_Predicted_Category_Count
```

```python
MultiGram_Stopwords, MultiGramTokens_After_Removing_Stop_Words_Training = create_stop_words_and_unique_tokens(MultiGramTokens_Train, MultiGramTokens_Test, top_n=200)
```

```python
Voc_M_Size = len(MultiGramTokens_After_Removing_Stop_Words_Training)
```

```python
MultiGram_Token_For_Each_Category, MultiGram_Each_Token_Count = count_each_category_tokens(Training_data)
```

```python
MultiGram_Liklihood_For_Each_Token = calculate_likelihood(MultiGram_Each_Token_Count, MultiGramTokens_After_Removing_Stop_Words_Training, MultiGram_Token_For_Each_Category, Voc_M_Size)
```

```python
MultiGram_Accuracy, MultiGram_True_Predicted_Category, MultiGram_True_Predicted_Category_Count, MultiGram_Total_Predicted_Category_Count = predict_category(Testing_data, Prior_Probability, MultiGram_Liklihood_For_Each_Token, MultiGram_Stopwords, Multigram)
```
